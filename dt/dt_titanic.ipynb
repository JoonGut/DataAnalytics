## Ejercicio 3: Filtrado básico

Aplica filtros para encontrar información específica:

1. ¿Cuántos pasajeros sobrevivieron? (Survived = 1)
2. ¿Cuántas mujeres había en el barco?
3. ¿Cuántos pasajeros de primera clase había?
4. Encuentra todos los pasajeros menores de 18 años
5. ¿Cuántos pasajeros pagaron más de 100 en tarifa?
6. Muestra los nombres de los primeros 5 pasajeros menores de 18 años

```python
# Tu código aquí

# 1. Pasajeros que sobrevivieron
sobrevivientes = titanic['Survived'].value_counts()[1]
print("Sobrevivieron: ",sobrevivientes," pasajeros")
# 2. Mujeres en el barco
mujeres = titanic['Sex'].value_counts()['female']
print("El total de mujeres en el titanic es de : ",mujeres)
# 3. Pasajeros de primera clase
pasajeros_primera = titanic['Pclass'].value_counts()[1]
print("Pasajeros de primera clase: ",pasajeros_primera)
# 4. Menores de 18 años
pasajeros_menores = titanic[titanic['Age']<18]
print("Pasajeros menores de 18 años: ",len(pasajeros_menores))
# 5. Tarifa mayor a 100
titanic['Fare'] = (
   titanic['Fare']
    .astype(str)
    .str.replace('.' , '') 
    )
titanic ['Fare'] = pd.to_numeric (titanic['Fare'])
tarifas_altas = len(titanic[titanic['Fare'] > 100])
print ("Pasajeros con tarifa mayor a 100: ",tarifas_altas)
# 6. Nombres de los primeros 5 menores de 18
print("Los nombres de los primeros 5 pasajeros menores de 18 años son: ", pasajeros_menores['Name'].head(5))
```

## Ejercicio 4: Filtrado avanzado

Combina múltiples condiciones:

1. Mujeres que sobrevivieron
2. Hombres de primera clase que no sobrevivieron
3. Pasajeros entre 20 y 40 años (inclusive) que sobrevivieron
4. Pasajeros que embarcaron en 'S' o 'C' (usa isin())
5. Pasajeros cuyo nombre contiene 'Mrs.' (usa str.contains())
6. ¿Cuál fue la tasa de supervivencia de las mujeres vs los hombres?

```python
# Tu código aquí

# 1. Mujeres que sobrevivieron
mujeres_sobrevivientes = titanic[(titanic['Sex'] == 'female') & (titanic['Survived'] == 1)]
print("Mujeres que sobrevivieron: ",len(mujeres_sobrevivientes))
# 2. Hombres de primera clase que no sobrevivieron
hombres_primera_no_sobrevivientes = titanic[(titanic['Sex'] == 'male') & (titanic['Pclass'] == 1) & (titanic['Survived'] == 0)]
print("Hombres de primera clase que no sobrevivieron: ",len(hombres_primera_no_sobrevivientes))
# 3. Edad 20-40 que sobrevivieron
jovenes_sobrevivientes = titanic[(titanic['Age']>=20)  & (titanic['Age']<=40) & (titanic['Survived']==1)]
print("Sobrevivientes entre 20 y 40 años: ",len(jovenes_sobrevivientes))
# 4. Embarcaron en 'S' o 'C'
embarcaron_sc = titanic[(titanic['Embarked']=='S') | (titanic['Embarked']=='C')]
print("Pasajeros en clase S o C: ",len(embarcaron_sc))
# 5. Nombres con 'Mrs.'
nombre_mrs = titanic[titanic['Name'].str.contains('Mrs.')]
print("Nombres que contiene Mrs",len(nombre_mrs))
# 6. Tasa de supervivencia por sexo
tada_supervivencia_sexo = titanic.groupby('Sex')['Survived'].mean()*100
print("Tasa de supervivencia por sexo:\n",tada_supervivencia_sexo)
```

## Ejercicio 5: Estadísticas y agrupaciones

Calcula estadísticas descriptivas:

1. ¿Cuál es la edad promedio de todos los pasajeros?
2. ¿Cuál es la tarifa más alta y más baja pagada?
3. ¿Cuántos pasajeros hay de cada clase? (value_counts)
4. ¿Cuál es la edad promedio por clase? (groupby)
5. ¿Cuál es la tarifa promedio por puerto de embarque?
6. Crea una tabla que muestre la cantidad de sobrevivientes y no sobrevivientes por sexo

```python
# Tu código aquí

# 1. Edad promedio
edad_promedio = titanic['Age'].mean()
print("Edad promedio de los pasajeros: ",round(edad_promedio,2))
# 2. Tarifa máxima y mínima
tarifa_maxima = titanic['Fare'].max()
tarifa_minima = titanic['Fare'].min()
print("La tarifa máxima es: ",tarifa_maxima)
print("La tarifa mínima es: ",tarifa_minima)
# 3. Pasajeros por clase
pasajero_clase = (titanic['Pclass']).value_counts()
print("Los pasajeros por las diferentes clases son\n",pasajero_clase)
# 4. Edad promedio por clase
edad_pomedio = titanic.groupby('Pclass')['Age'].mean()
print("Edad promedio por clase:\n",edad_pomedio)
# 5. Tarifa promedio por puerto
titanic['Fare'] = (
   titanic['Fare']
    .astype(str)
    .str.replace('.' , '') 
    )
titanic ['Fare'] = pd.to_numeric (titanic['Fare'])
promedio_por_puerto = titanic.groupby('Embarked')['Fare'].mean()
print("Tarifa promedio por puerto:\n",round(promedio_por_puerto,2))
# 6. Supervivencia por sexo
supervivencia_por_sexo = titanic.groupby('Sex')['Survived'].mean()*100
print("Tasa de supervivencia por sexo:\n",supervivencia_por_sexo)
#crosstab
pd.crosstab(titanic['Sex'],titanic['Survived'])
```

## Ejercicio 6: Crear nuevas columnas

Manipula el DataFrame creando nuevas columnas:

1. Crea una columna 'Family_Size' que sume SibSp + Parch + 1
2. Crea una columna 'Age_Group' con las categorías:
   - 'Child' (< 18 años)
   - 'Adult' (18-59 años)
   - 'Senior' (>= 60 años)
   - 'Unknown' (edad faltante)
3. Crea una columna 'Fare_Category' con:
   - 'Low' (Fare < 10)
   - 'Medium' (10 <= Fare < 50)
   - 'High' (Fare >= 50)
4. Crea una columna 'Title' que extraiga el título del nombre (Mr., Mrs., Miss., etc.)
5. Muestra las primeras 10 filas con las nuevas columnas

```python
# Crear una copia para no modificar el original
titanic_trabajo = titanic.copy()

# Tu código aquí

# 1. Family_Size
titanic_trabajo['Family_Size'] = titanic_trabajo['SibSp'] + titanic_trabajo['Parch'] + 1
titanic_trabajo.head()
# 2. Age_Group
titanic_trabajo['Age_Group'] = pd.cut(
    titanic_trabajo['Age'],
    bins=[0, 18, 59, 120],
    labels=['Child', 'Adult', 'Senior']
)
# Añadimos a mano el 'Unknown' si es nulo, en esto caso los nulos son 'Unknown'
titanic_trabajo['Age_Group'] = titanic_trabajo['Age_Group'].cat.add_categories('Unknown').fillna('Unknown')
titanic_trabajo.head()


```

```python
# 3. Fare_Category
titanic['Fare'] = (
   titanic['Fare']
    .astype(str)
    .str.replace('.' , '') 
    )
titanic ['Fare'] = pd.to_numeric (titanic['Fare'])
titanic_trabajo['Fare_Category'] = pd.cut(
    titanic['Fare'],
    bins=[-1, 10, 50, float('inf')],
    labels=['Low', 'Medium', 'High']
)
titanic_trabajo.head()
```

```python
# 4. Title (pista: usa str.extract con regex)
titanic_trabajo['Title'] = titanic_trabajo['Name'].str.extract(' ([A-Za-z]+)\.')
# 5. Mostrar primeras 10 filas con nuevas columnas
titanic_trabajo.head(10)
```

## Ejercicio 7: Análisis de valores faltantes

Analiza los datos faltantes en el dataset:

1. ¿Qué columnas tienen valores faltantes y cuántos?
2. ¿Qué porcentaje de valores faltantes tiene cada columna?
3. ¿Cuántas filas tienen al menos un valor faltante?
4. Muestra 5 ejemplos de filas que tienen edad faltante
5. ¿Hay algún patrón en los valores faltantes de edad por clase?

```python
# Tu código aquí

# 1. Valores faltantes por columna
faltantes_por_columna = titanic_trabajo.isnull().sum()
print("Valores faltantes por columna:\n",faltantes_por_columna)
# 2. Porcentaje de valores faltantes
faltantes_porcentaje = titanic_trabajo.isnull().mean() * 100
print("\nPorcentaje de valores faltantes por columna (%):\n", faltantes_porcentaje.round(2))
```

```python
# 3. Filas con al menos un valor faltante
faltantes_filas = titanic_trabajo.isnull().any(axis=1)
faltantes_filas_sum = faltantes_filas.sum()
print("\nNúmero de filas con al menos un valor faltante:", faltantes_filas_sum)

 
# 4. Ejemplos de filas con edad faltante
ejemplos_edad_faltante = titanic_trabajo[titanic_trabajo['Age'].isnull()]
print("\nEjemplos de filas con edad faltante:\n", ejemplos_edad_faltante)
```

```python
# 5. Patrón de edad faltante por clase 
titanic_trabajo['Title'] = titanic_trabajo['Name'].str.extract(r',\s*([^\.]*)\.')

titanic_trabajo['AgeMissing'] = titanic_trabajo['Age'].isna()

patron = titanic_trabajo.groupby('Pclass')['AgeMissing'].mean() * 100

print("\nPorcentaje de edades faltantes por clase:\n",patron.round(2))
```

## Ejercicio 8: Análisis avanzado

Realiza un análisis más profundo:

1. ¿Cuál fue la tasa de supervivencia por clase y sexo? (groupby con múltiples columnas)
2. ¿Cuáles fueron los 5 pasajeros que pagaron las tarifas más altas?
3. ¿Cuáles fueron los 5 sobrevivientes más jóvenes?
4. ¿Qué títulos tenían mayor tasa de supervivencia? (usa la columna Title del ejercicio 6)
5. ¿Cómo se relaciona el tamaño de familia con la supervivencia?
6. Ordena las clases por tarifa promedio de mayor a menor

```python
# Tu código aquí

# 1. Supervivencia por clase y sexo
supervivencia_clase_sexo = titanic_trabajo.groupby(['Pclass', 'Sex'])['Survived'].mean() * 100
print("\nSupervivencia por clase y sexo (%):\n", supervivencia_clase_sexo)

# 2. Top 5 tarifas más altas
titanic['Fare'] = (
   titanic['Fare']
    .astype(str)
    .str.replace('.' , '') 
    )
titanic ['Fare'] = pd.to_numeric (titanic['Fare'])
top5_fares = titanic['Fare'].nlargest(5)
print("Top 5 tarifas más altas:\n", top5_fares)

# 3. Top 5 sobrevivientes más jóvenes
supervivientes_jovenes = titanic_trabajo.sort_values(by='Age', ascending=True).head(5)
print("\nTop 5 sobrevivientes más jóvenes\n",supervivientes_jovenes)

```

```python
# 4. Supervivencia por título
supervivencia_titulo = titanic_trabajo.groupby('Title')['Survived'].mean() * 100
print("\nSupervivencia por título (%):\n", supervivencia_titulo)

# 5. Tamaño de familia vs supervivencia
fam_superv = titanic_trabajo.groupby('Family_Size')['Survived'].agg(['mean','count'])
fam_superv['survival_rate_pct'] = fam_superv['mean'] * 100
print("\nTamaño de familia vs supervivencia (% y conteo):\n", fam_superv[['survival_rate_pct','count']])

# 6. Clases ordenadas por tarifa promedio
clase_tarifa = titanic_trabajo.groupby('Pclass')['Fare'].mean().sort_values(ascending=False)
print("\nClases ordenadas por tarifa promedio (de mayor a menor):\n", clase_tarifa)
```
