```python
import pandas as pd
premier = pd.read_csv('premier.csv')
premier.info()
```

### 1️⃣ Media de edad de los jugadores del Arsenal


Calcula la **edad media** de los jugadores cuyo club sea **Arsenal**.


```python
edad_arsenal = premier.loc[premier['Club'] == 'Arsenal', 'Age'].mean()
print("La edad media de los jugadores del Arsenal es: ",edad_arsenal)
```

### 2️⃣ Jugador del Liverpool con más y menos goles


Muestra el **jugador del Liverpool** que haya marcado **más goles** y el que haya marcado **menos goles**.


```python
# Escribe tu código aquí
datosLiverpool = premier[premier['Club']== 'Liverpool']
mayorGoleador = datosLiverpool.sort_values('Goals',ascending=False).head(1)['Name']
menorGoleador = datosLiverpool.sort_values('Goals',ascending=True).head(1)['Name']
print("El mayor goleador del Liverpool es: \n",mayorGoleador)
print("\nEl menor goleador del Liverpool es: \n",menorGoleador)



```

### 3️⃣ Equipos con más tarjetas rojas


Calcula el **número total de tarjetas rojas por club** (`Red cards`).  
Ordena los resultados de mayor a menor y muestra los **5 equipos con más expulsiones**.


```python
# Escribe tu cimport pandas as pd
equipos_tarjetas = premier.groupby('Club')['Red cards'].sum()
tarjetas_ordenadas = equipos_tarjetas.sort_values(ascending=False).head(5)

print("Los 5 equipos con mas expulsiones son:\n",tarjetas_ordenadas)
```

### 4️⃣ Nacionalidades


- ¿Cuántas **nacionalidades diferentes** hay en la liga?  
- ¿Cuál es la **nacionalidad más frecuente**?


```python
# Escribe tu código aquí
nacionalidades_diferentes = premier['Nationality'].nunique()
print ("En la premier tenemos ",nacionalidades_diferentes," nacionalidades diferentes")
```

```python
nacionalidad_frecuente = premier['Nationality'].value_counts()
print("La nacionalidad mas frecuente es:\n",nacionalidad_frecuente.head(1))
```

### 5️⃣ Top 10 jugadores con más asistencias


Muestra los **10 jugadores con mayor asistencias** (`Assits`).  
Incluye las columnas: `Name`, `Club`, y `Assits`.


```python
# Top 10 jugadores con más asistencias
copia_premier = premier.copy()
top10_asistencias = copia_premier[['Name', 'Club', 'Assists']].sort_values('Assists', ascending=False).head(10)
print(top10_asistencias)
```

### 6️⃣ Clasificación de jugadores por goles


Crea una nueva columna llamada `Rango_goles` que clasifique a los jugadores según sus goles (`Goals`):  
- **Sin gol** → 0 goles  
- **Pocos goles** → entre 1 y 4 goles  
- **Goleador** → 5 o más goles  

Después, muestra cuántos jugadores hay en cada categoría.


```python
# Escribe tu código aquí
premier['Rango Goles'] = pd.cut(
    premier['Goals'],
    bins = [-1,0,4,premier['Goals'].max()],
    labels = ['Sin Gol', 'Pocos Goles', 'Goleador'],
)
premier['Rango Goles'].value_counts()


```

### 7️⃣ Tabla pivote con media de goles y asistencias por posición


Crea una **tabla pivote** que muestre la **media de goles y asistencias** (`Goals`, `Assists`)  
por **posición de juego** (`Position`).


```python
# Escribe tu código aquí
pd.pivot_table(premier, 
                values=['Goals','Assists'],
                index="Position", 
                aggfunc=["mean"]
               ).round(2)
```
