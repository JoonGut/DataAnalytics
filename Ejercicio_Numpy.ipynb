## Ejercicio 3: Indexing y Slicing

Dado el array de temperaturas de una semana:

```python
temperaturas = np.array([18, 22, 25, 19, 27, 30, 24])
dias = ['Lun', 'Mar', 'Mié', 'Jue', 'Vie', 'Sáb', 'Dom']
```

Responde:
1. ¿Cuál fue la temperatura del miércoles?
2. ¿Cuáles fueron las temperaturas de los primeros 3 días?
3. ¿Cuáles fueron las temperaturas del fin de semana (sábado y domingo)?
4. ¿Qué días tuvieron temperatura superior a 25°?
5. ¿Cuál es la temperatura promedio de los días laborables (lunes a viernes)?

````python
# Datos
temperaturas = np.array([18, 22, 25, 19, 27, 30, 24])
dias = np.array(['Lun', 'Mar', 'Mié', 'Jue', 'Vie', 'Sáb', 'Dom'])

# Tu código aquí

# 1. Temperatura del miércoles (índice 2)
indice_dias = np.where(dias == 'Mié')
temp_miercoles = temperaturas[indice_dias]
print(temp_miercoles)
# 2. Temperaturas primeros 3 días
temp_primeros_3 = temperaturas[:3]
print(temp_primeros_3)
# 3. Temperaturas fin de semana
temp_finde = temperaturas[-2:]
print(temp_finde)
# 4. Días con temperatura > 25°
dias_temp_altas = temperaturas [temperaturas > 25]
print (f"Dias con temperaturas altas: {dias_temp_altas}")
# 5. Promedio días laborables (índices 0-4)
promedio_primeros_dias = np.mean(temperaturas[:4])
print (f"El promedio es: {round(promedio_primeros_dias,2)}")
````

## Ejercicio 4: Funciones Estadísticas

Las calificaciones de un examen fueron:

```python
calificaciones = np.array([85, 92, 78, 96, 88, 73, 89, 94, 67, 91])
```

Calcula:
1. La calificación promedio
2. La mediana
3. La calificación más alta y más baja
4. La desviación estándar
5. ¿Cuántos estudiantes obtuvieron una calificación superior al promedio?
6. ¿Cuál es el percentil 75 (tercer cuartil)?

````python
# Datos
calificaciones = np.array([85, 92, 78, 96, 88, 73, 89, 94, 67, 91])

# Tu código aquí

# 1. Promedio
calificacion_promedio = np.mean(calificaciones)
print (f"La calificacion promedio es: {calificacion_promedio}")
# 2. Mediana
calificacion_mediana = np.median(calificaciones)
print(f"La calificacion mediana es: {calificacion_mediana}")
# 3. Máxima y mínima
calificacion_max = np.max(calificaciones)
print(f"La calificacion maxima: {calificacion_max}")
calificacion_min = np.min(calificaciones)
print(f"La calificacion minima: {calificacion_min}")

# 4. Desviación estándar
desviacion_estandar = np.std(calificaciones)
print(f"Desviacion estandar, {desviacion_estandar}")
# 5. Estudiantes sobre el promedio
estudias_mejor_promedio = calificaciones[calificaciones > calificacion_promedio]
indices_mejor_promedio = np.where(calificaciones > calificacion_promedio)
print(f"Los índices de los estudiantes mejor al promedio son {indices_mejor_promedio[0]}")
# 6. Percentil 75
percentil_75 = np.percentile(calificaciones, 75)
print(f"El percentil 75 es: {percentil_75}")
````

## Ejercicio 5: Arrays 2D (Matrices)

Crea una matriz que represente las ventas de 3 productos durante 4 meses:

```
           Ene  Feb  Mar  Abr
Producto A: 120  150  180  200
Producto B:  80   90  110  130
Producto C: 200  220  250  280
```

Responde:
1. Crea la matriz de ventas
2. ¿Cuáles fueron las ventas del Producto B en marzo?
3. ¿Cuáles fueron las ventas totales de febrero (todos los productos)?
4. ¿Cuál producto tuvo las mayores ventas totales en el período?
5. ¿En qué mes se vendió más en total?
6. Calcula el promedio de ventas por producto
7. Calcula el promedio de ventas por mes

````python
# Tu código aquí

# 1. Crear matriz de ventas
matriz_ventas = np.array([
#   Ene   Feb   Mar  Abr
    [120, 150, 180, 200],   # Producto A
    [80, 90, 110, 130],     # Producto B
    [200, 220, 250, 280]    # Producto C
])
matriz_meses = np.array(['Ene', 'Feb', 'Mar', 'Abr'])
matriz_productos = np.array(['Producto_A', 'Producto_B', 'Producto_C'])

print("\t", matriz_meses)
for producto,ventas in zip(matriz_productos, matriz_ventas):
    print(f"{producto}: {ventas}")

# 2. Ventas Producto B en marzo
indice_marzo = np.where(matriz_meses == 'Mar')
indice_productoB = np.where(matriz_productos == 'Producto_B')
ventas_productoB = matriz_ventas[indice_productoB,indice_marzo]
#print(f"{indice_marzo}{indice_productoB}")
print(f"Las ventas del Producto B en marzo son: {ventas_productoB}")


# 3.Ventas totales de febrero
ventas_total_feb = sum(matriz_ventas[:][1])
print(f"Las ventas totales de febrero {ventas_total_feb}")

# 4. Producto con mayores ventas totales

producto_max_ventas = np.sum(matriz_ventas,axis=1)
indice_producto_max = np.argmax(producto_max_ventas)
print (f"El producto con mayores ventas es el {matriz_productos[indice_producto_max]}")

# 5. Mes con más ventas
ventas_por_mes = np.sum(matriz_ventas, axis=0)
indice_mes_max = np.argmax(ventas_por_mes)
print(f"El índice del mes con más ventas es: {matriz_meses[indice_mes_max]}")

# 6. Promedio por producto
promedio_producto = np.mean(matriz_ventas,axis=1)
for promedio,producto in zip (promedio_producto,matriz_productos):
    print(f"El producto: {producto} tiene un promedio de: {round(promedio,2)}")
# 7. Promedio por mes
promedio_mes = np.mean(matriz_ventas,axis=0)
for promedio,mes in zip (promedio_mes,matriz_meses):
    print(f"En el mes: {mes} tiene el promedio: {round(promedio,2)}")
````

## Ejercicio 6: Análisis de Datos de Sensores

Un sensor registra temperaturas cada hora durante un día. Los datos fueron:

```python
temperaturas_sensor = np.array([15.2, 16.1, 17.3, 19.2, 22.1, 25.4, 28.3, 30.1, 
                               31.5, 32.2, 33.1, 31.8, 30.5, 28.9, 26.7, 24.3, 
                               22.1, 20.5, 18.9, 17.2, 16.1, 15.8, 15.3, 15.1])
```

Realiza el siguiente análisis:
1. ¿Cuál fue la temperatura máxima y a qué hora ocurrió?
2. ¿Cuál fue la temperatura mínima y a qué hora ocurrió?
3. Calcula la temperatura promedio del día
4. ¿En cuántas horas la temperatura estuvo por encima de 25°?
5. Calcula la diferencia entre la temperatura máxima y mínima
6. Encuentra las horas donde la temperatura estuvo entre 20° y 25° (inclusive)

````python
# Datos
temperaturas_sensor = np.array([15.2, 16.1, 17.3, 19.2, 22.1, 25.4, 28.3, 30.1, 
                               31.5, 32.2, 33.1, 31.8, 30.5, 28.9, 26.7, 24.3, 
                               22.1, 20.5, 18.9, 17.2, 16.1, 15.8, 15.3, 15.1])

# Tu código aquí

# 1. Temperatura máxima y hora
indice_temp_max = np.argmax(temperaturas_sensor)
print (f"La temperatura maxima alcanzada fue {temperaturas_sensor[indice_temp_max]} a las {indice_temp_max}")

# 2. Temperatura mínima y hora
indice_temp_min = np.argmin(temperaturas_sensor)
print(f"La temperatura minima alcanzada fue {temperaturas_sensor[indice_temp_min]} a las {indice_temp_min}")

# 3. Temperatura promedio
temp_promedio = np.mean(temperaturas_sensor)
print(f"Temperatura promedio del día: {temp_promedio}°C")

# 4. Horas con temperatura > 25°
indices_temp_alta = np.where(temperaturas_sensor > 25)
print(f"Las horas con temperaturas superiores a 25°C son: {indices_temp_alta[0]}")

# 5. Diferencia máxima-mínima
diferencia_max_min = max(temperaturas_sensor) - min(temperaturas_sensor)
print(f"Diferencia entre temperatura máxima y mínima: {diferencia_max_min}°C")

# 6. Horas con temperatura entre 20° y 25° 
horas_temp_media = np.where((temperaturas_sensor >= 20) & (temperaturas_sensor <= 25))[0]
print (f"Las horas con temperaturas agradebles son las siguientes: {horas_temp_media}") 
````

## Ejercicio 7: Análisis Financiero

Una empresa tiene los siguientes datos trimestrales de ingresos y gastos:

```python
ingresos = np.array([50000, 65000, 70000, 80000])  # Q1, Q2, Q3, Q4
gastos = np.array([35000, 45000, 50000, 55000])   # Q1, Q2, Q3, Q4
```

Calcula:
1. El beneficio por trimestre (ingresos - gastos)
2. El beneficio total del año
3. El trimestre con mayor beneficio
4. El margen de beneficio por trimestre (beneficio/ingresos * 100)
5. ¿En qué trimestres el margen fue superior al 25%?
6. La tasa de crecimiento de ingresos trimestre a trimestre (`Utilizar funcion div`)
7. El promedio de gastos del año

````python
# Datos
ingresos = np.array([50000, 65000, 70000, 80000])  # Q1, Q2, Q3, Q4
gastos = np.array([35000, 45000, 50000, 55000])   # Q1, Q2, Q3, Q4
trimestres = np.array(['Q1', 'Q2', 'Q3', 'Q4'])

# Tu código aquí

# 1. Beneficio por trimestre
beneficios_trimestre = ingresos - gastos
#print (beneficios_trimestre)
for trimestre, beneficio_trimestre in zip (trimestres, beneficios_trimestre):
    print(f"En los trimestres {trimestre}, un beneficio de: {beneficio_trimestre}€")

# 2. Beneficio total del año
beneficio_anio = sum(ingresos) - sum(gastos)
print (f"El beneficio total es de: {beneficio_anio}€")
# 3. Trimestre con mayor beneficio
beneficios_mx_trimestre = np.argmax(beneficios_trimestre)
print(beneficios_mx_trimestre)
print (f"El mejor trimestre {trimestres[beneficios_mx_trimestre]} tuvo un beneficio de: {beneficios_trimestre[beneficios_mx_trimestre]}")

# 4. Margen de beneficio por trimestre (%)
margenes_beneficio = ( beneficios_trimestre / ingresos ) * 100
for trimestre, margen_beneficio in zip(trimestres, margenes_beneficio):
    print(f"En los trimestres {trimestre} hay un margen de: {round(margen_beneficio,2)} %")

# 5. Trimestres con margen > 25%
margenes_alto = np.where(margenes_beneficio > 25)
print(f"Los trimestres {trimestres[margenes_alto]} superan el margen del 25 % ")

# 6. Tasa de crecimiento de ingresos
tasa_crecimiento = np.divide(ingresos[1:], ingresos[:-1]) - 1
print(tasa_crecimiento)
# 7. Promedio de gastos
promedio_gastos = np.mean(gastos)
print(f"El promedio de gastos es: {promedio_gastos}€")
````
